```python
# import libraries
import pandas as pd
import numpy as np
```

```python
# remove unnecessary column
df=df.drop(columns=['Transaction ID'])
```

```python
# correct data types
num_cols = ['Quantity','Price Per Unit','Total Spent']
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
```

```python
# convert transaction date
df['Transaction Date']= pd.to_datetime(df['Transaction Date'],errors = 'coerce')
```

```python
# Handle missing values in item
df=df.dropna(subset=['Item'])
```

```python
print(df['Item'].isnull().sum())
print(df.shape)
```

```python
#Handle Missing numerical values
df['Quantity']=df['Quantity'].fillna(df['Quantity'].median())
df['Price Per Unit']=df['Price Per Unit'].fillna(df['Price Per Unit'].median())
df['Total Spent']=df['Total Spent'].fillna(df['Quantity']*df['Price Per Unit'])
```

```python
#handle missing categorical values
df['Payment Method'] = df['Payment Method'].replace(['UNKNOWN','Unknown',''],pd.NA)
df['Location'] = df['Location'].replace(['UNKNOWN','Unknown',''],pd.NA)
df['Payment Method'] = df['Payment Method'].fillna('Unknown')
df['Location'] = df['Location'].fillna('Unknown')
```

```python
# Validate the clean Dataset
print(df.shape)
print(df.isnull().sum())
print(df.dtypes)
print(df.duplicated().sum())
print(df.head())
```
